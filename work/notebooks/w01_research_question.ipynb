## 3. Quick look at the data (2-3 real numbers)

*Load the starter CSV below and show 2-3 real numbers that make your lane look worth the next 7 weeks.*

```python
import pandas as pd

df = pd.read_csv("data/raw/content_refresh_anonymized.csv")

df.head()
```

```python
ctr_by_position = df.groupby("position_tier")["ctr"].mean().round(3)
print(ctr_by_position)
```

CTR differs across ranking positions, indicating that pages have different visibility and engagement patterns.

```python
trend_counts = df["trend_direction"].value_counts()
print(trend_counts)
```

The dataset contains multiple trend categories, showing that page performance varies across the content inventory.

```python
word_count_by_trend = df.groupby("trend_direction")["word_count"].median().round(0)
print(word_count_by_trend)
```

Growing and declining pages have similar median word counts, suggesting that content length alone may not explain performance differences.
